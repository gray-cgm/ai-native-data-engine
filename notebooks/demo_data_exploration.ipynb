# AI Data Loop Engine Demo Exploration

This notebook inspects the local DuckDB and Parquet outputs produced by the project workflows.

```python
from pathlib import Path
import duckdb
import pandas as pd

repo_root = Path.cwd()
duckdb_path = repo_root / 'data' / 'duckdb' / 'app.duckdb'
silver_path = repo_root / 'data' / 'silver' / 'samples.parquet'
gold_path = repo_root / 'data' / 'gold' / 'dataset_samples.parquet'

print('duckdb exists:', duckdb_path.exists(), duckdb_path)
print('silver parquet exists:', silver_path.exists(), silver_path)
print('gold parquet exists:', gold_path.exists(), gold_path)
```

```python
con = duckdb.connect(str(duckdb_path))
con.execute("show tables").df()
```

```python
distribution_df = con.execute("select scene, count(*) as sample_count from samples group by scene order by sample_count desc").df()
distribution_df
```

```python
samples_df = con.execute("select * from samples order by timestamp").df()
samples_df.head()
```

```python
if gold_path.exists():
    gold_df = duckdb.sql(f"select * from read_parquet('{gold_path.as_posix()}')").df()
    display(gold_df.head())
else:
    print('Run ingestion/materialization first to create data/gold/dataset_samples.parquet')
```
